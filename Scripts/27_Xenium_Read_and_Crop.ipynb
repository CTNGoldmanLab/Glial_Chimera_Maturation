## Use interactive napari session to subset the data to regions with engrafted human cells

### WA09_GFP_44471

```python
# Start napari-spatialdata interactive session to select the regions of interest
%matplotlib inline

plt.rcParams['figure.figsize'] = (20, 20)

interactive = Interactive(sdata_WA09_GFP_44471)
interactive.run()
```

```python
# check that the shapes layer was saved correctly
print(sdata_WA09_GFP_44471)

# visualize the ROIs on the tissue and check to make sure that the cell boundaries and ROI polygons are in the same coordinate system
sdata_WA09_GFP_44471.pl.render_shapes("cell_boundaries").pl.render_shapes("human_cells", color="lightblue").pl.show()
```

```python
# had trouble getting the 'table' output after subsetting the polygons, so I used this method to update the table in the SpatialData object. 

# first, set the anndata index to cell_id
adata_WA09_GFP_44471 = sdata_WA09_GFP_44471["table"]
adata_WA09_GFP_44471.obs = adata_WA09_GFP_44471.obs.set_index("cell_id", drop=False)
sdata_WA09_GFP_44471.tables["table"] = adata_WA09_GFP_44471  # Update the table in your SpatialData object

# ensure anndata index matches the spatial element index
print(adata_WA09_GFP_44471.obs.index[:5])
print(sdata_WA09_GFP_44471["cell_boundaries"].index[:5])
```

```python
# remove any existing spatialdata_attrs metadata
adata_WA09_GFP_44471 = sdata_WA09_GFP_44471["table"]
if "spatialdata_attrs" in adata_WA09_GFP_44471.uns:
    del adata_WA09_GFP_44471.uns["spatialdata_attrs"]
```

```python
#parse the anndata to set all required metadata, allows spatialdata to recognize the linkages between the anndata and the spatial elements. 
from spatialdata.models import TableModel

adata_WA09_GFP_44471_parsed = TableModel.parse(
    adata_WA09_GFP_44471,
    region="cell_boundaries",
    region_key="region",
    instance_key="cell_id"
)

sdata_WA09_GFP_44471.tables["table"] = adata_WA09_GFP_44471_parsed
```

```python
# Now ready for subsetting based on ROIs
# Extract cells under each of the regions of interest
# following this tutorial: https://spatialdata.scverse.org/en/latest/tutorials/notebooks/notebooks/examples/napari_rois.html#extracting-cells-under-the-regions

from spatialdata import polygon_query

roi_shapes = {
    "human_cells": sdata_WA09_GFP_44471["human_cells"].geometry.iloc[0]
}

filtered_tables = {}
for roi_name, polygon in roi_shapes.items():
    result = polygon_query(sdata_WA09_GFP_44471, polygon=polygon, target_coordinate_system="global")
    if "table" in result.tables:
        filtered_tables[roi_name] = result["table"]
    else:
        filtered_tables[roi_name] = None  # No cells in this ROI

filtered_tables
```

```python
# now annotate the cells in the larger sdata table with the ROI names, or 'unassigned' if they don't fall into any of the ROIs. 
roi_to_group = {
    "human_cells": "human_cells"
}

# Ensure the index is unique
assert sdata_WA09_GFP_44471["table"].obs.index.is_unique

# Use only unique group names for categories
categories = ["unassigned"] + sorted(set(roi_to_group.values()))
n = len(sdata_WA09_GFP_44471["table"])

# Initialize the annotation column
sdata_WA09_GFP_44471["table"].obs["annotation"] = pd.Categorical(["unassigned"] * n, categories=categories)

# Assign group names to cells in each ROI
for roi_name, subtable in filtered_tables.items():
    if subtable is not None:
        in_roi = subtable.obs.index
        group = roi_to_group[roi_name]
        sdata_WA09_GFP_44471["table"].obs.loc[in_roi, "annotation"] = group

# Check the annotation counts
print(sdata_WA09_GFP_44471["table"].obs["annotation"].value_counts())
print(sdata_WA09_GFP_44471)
```

```python
# Save the annotated SpatialData object
zarr_path_WA09_GFP_44471_human = f"{ZARR_DIR}/xenium_zarr_path_WA09_GFP_44471_human.zarr"

sdata_WA09_GFP_44471.write(zarr_path_WA09_GFP_44471_human)
```
